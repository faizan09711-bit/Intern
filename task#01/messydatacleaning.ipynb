```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = pd.read_csv(r"Z:\Intermship\us-shein-beauty_and_health-4267.csv")
data.head(5000)
```

```python
print(data.isnull().sum())
data_before=data
```

```python
datatype=data['price'].dtype
print(datatype)
data['price']=data['price'].str.replace('$','').astype(float)
print(data['price'].dtype)
data.head()
```

```python
data['discount'] = data['discount'].str.replace('%', '').astype(float)
print(data['discount'].dtype)
data['discount']=data['discount'].fillna(data['discount'].median())
data.head()
```

```python
print(data['color-count'].dtype)
data['color-count'] = data['color-count'].fillna(data['color-count'].median())
print(data['color-count'].dtype)
data.head()
```

```python
data2=data[['rank-title','rank-sub','goods-title-link','goods-title-link--jump','goods-title-link--jump href','selling_proposition']]
for i in data2:
    data[i] = data[i].fillna(data[i].mode()[0])
data.head()

```

```python
data2.head()
```

```python

#print(data_before.isnull().sum())
print(data.isnull().sum())

g = data.groupby('rank-title') 

min_size = g.size().min()
data = g.apply(lambda x: x.sample(min_size)).reset_index(drop=True)
 
data.head()
```
