```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
data = pd.read_csv(r"Z:\Intermship\task#01\restaurant_sales_dataset.csv")
data.head()
print(data.isnull().sum())


print(data[['Order ID','Customer ID','Category','Item','Price','Quantity','Order Total','Order Date','Payment Method']].dtypes)



```

```python
data_before=data
```

```python
data['Item']=data['Item'].astype(str)
data['Item']=data['Item'].fillna(data['Item'].mode()[0])
data['Payment Method']=data['Payment Method'].astype(str)
data['Payment Method']=data['Payment Method'].fillna(data['Payment Method'].mode()[0])
```

```python
# data['Item']=data['Item'].fillna(data['Item'].median())
# data['Payment Method']=data['Payment Method'].fillna(data['Payment Method'].median())
print(data.isnull().sum())
```

```python
num=data[['Order Total','Quantity','Price']]
print(num.dtypes)
for i in num:
    data[i]=data[i].fillna(data[i].median())
print(data.isnull().sum())

```

```python
g = data.groupby('Order Total') 
min_size = g.size().min()
data = g.apply(lambda x: x.sample(min_size)).reset_index(drop=True)
```
